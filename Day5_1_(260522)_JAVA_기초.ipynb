```python
%%writefile /content/ConditionExample.java
public class ConditionExample {
    public static void main(String[] args) {
        int score = 78;

        if (score >= 90) {
            System.out.println("A등급");
        } else if (score >= 80) {
            System.out.println("B등급");
        } else if (score >= 70) {
            System.out.println("C등급");
        } else {
            System.out.println("D등급");
        }
    }
}
```

```python
%%writefile /content/LoopExample.java
public class LoopExample {
    public static void main(String[] args) {
        int total = 0;

        for (int i = 1; i <= 100; i++) {
            total = total + i;
        }

        System.out.println("1부터 100까지의 합계: " + total);
    }
}
```

```python
%%writefile /content/EvenOddTotalExample.java
public class EvenOddTotalExample {
    public static void main(String[] args) {
        int evenTotal = 0;
        int oddTotal = 0;

        for (int i = 1; i <= 10; i++) {
            if (i % 2 == 0) {
                evenTotal = evenTotal + i;
            } else {
                oddTotal = oddTotal + i;
            }
        }

        System.out.println("짝수 합계: " + evenTotal);
        System.out.println("홀수 합계: " + oddTotal);
    }
}
```

```python
%%writefile /content/StudentScoreExample.java
public class StudentScoreExample {
    public static void main(String[] args) {
        String name = "이자바";
        int korean = 85;
        int english = 90;
        int math = 78;

        int total = korean + english + math;
        double average = total / 3.0;

        System.out.println("학생명: " + name);
        System.out.println("국어: " + korean);
        System.out.println("영어: " + english);
        System.out.println("수학: " + math);
        System.out.println("총점: " + total);
        System.out.println("평균: " + average);
    }
}
```
