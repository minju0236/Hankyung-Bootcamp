```python
%%writefile spring-lab/finance-security-app/src/main/java/com/example/demo/repository/UserRepository.java

package com.example.demo.repository;

import com.example.demo.domain.AppUser;
import com.example.demo.dto.SignupForm;
import org.springframework.dao.EmptyResultDataAccessException;
import org.springframework.jdbc.core.JdbcTemplate;
import org.springframework.jdbc.core.RowMapper;
import org.springframework.stereotype.Repository;

import java.util.List;
import java.util.Optional;

@Repository
public class UserRepository {

    private final JdbcTemplate jdbcTemplate;

    public UserRepository(JdbcTemplate jdbcTemplate) {
        this.jdbcTemplate = jdbcTemplate;
    }

    private final RowMapper<AppUser> userRowMapper = (rs, rowNum) -> new AppUser(
            rs.getLong("id"),
            rs.getString("email"),
            rs.getString("password"),
            rs.getString("name"),
            rs.getString("role"),
            rs.getString("status"),
            rs.getTimestamp("created_at").toLocalDateTime()
    );

    public Optional<AppUser> findByEmail(String email) {
        String sql = """
                SELECT id, email, password, name, role, status, created_at
                FROM users
                WHERE email = ?
                """;

        try {
            AppUser user = jdbcTemplate.queryForObject(sql, userRowMapper, email);
            return Optional.of(user);
        } catch (EmptyResultDataAccessException e) {
            return Optional.empty();
        }
    }

    public Optional<AppUser> findById(Long id) {
        String sql = """
                SELECT id, email, password, name, role, status, created_at
                FROM users
                WHERE id = ?
                """;

        try {
            AppUser user = jdbcTemplate.queryForObject(sql, userRowMapper, id);
            return Optional.of(user);
        } catch (EmptyResultDataAccessException e) {
            return Optional.empty();
        }
    }

    public List<AppUser> findAll() {
        String sql = """
                SELECT id, email, password, name, role, status, created_at
                FROM users
                ORDER BY id DESC
                """;

        return jdbcTemplate.query(sql, userRowMapper);
    }

    public void save(SignupForm form, String encodedPassword) {
        String sql = """
                INSERT INTO users (email, password, name, role, status)
                VALUES (?, ?, ?, 'USER', 'ACTIVE')
                """;

        jdbcTemplate.update(
                sql,
                form.getEmail(),
                encodedPassword,
                form.getName()
        );
    }

    public void updateRole(Long userId, String role) {
        String sql = """
                UPDATE users
                SET role = ?
                WHERE id = ?
                """;

        jdbcTemplate.update(sql, role, userId);
    }

    public void updateStatus(Long userId, String status) {
        String sql = """
                UPDATE users
                SET status = ?
                WHERE id = ?
                """;

        jdbcTemplate.update(sql, status, userId);
    }
}
```

```python
%%writefile spring-lab/finance-security-app/src/main/java/com/example/demo/repository/AccountTransactionRepository.java

package com.example.demo.repository;

import com.example.demo.domain.AccountTransaction;
import org.springframework.jdbc.core.JdbcTemplate;
import org.springframework.jdbc.core.RowMapper;
import org.springframework.stereotype.Repository;

import java.util.List;

@Repository
public class AccountTransactionRepository {

    private final JdbcTemplate jdbcTemplate;

    public AccountTransactionRepository(JdbcTemplate jdbcTemplate) {
        this.jdbcTemplate = jdbcTemplate;
    }

    private final RowMapper<AccountTransaction> transactionRowMapper = (rs, rowNum) -> new AccountTransaction(
            rs.getLong("id"),
            rs.getLong("user_id"),
            rs.getLong("account_id"),
            rs.getObject("target_account_id") == null ? null : rs.getLong("target_account_id"),
            rs.getString("account_number"),
            rs.getString("target_account_number"),
            rs.getString("transaction_type"),
            rs.getLong("amount"),
            rs.getString("memo"),
            rs.getLong("balance_after"),
            rs.getTimestamp("created_at").toLocalDateTime()
    );

    public List<AccountTransaction> findByUserId(Long userId) {
        String sql = """
                SELECT
                    t.id,
                    t.user_id,
                    t.account_id,
                    t.target_account_id,
                    a.account_number AS account_number,
                    ta.account_number AS target_account_number,
                    t.transaction_type,
                    t.amount,
                    t.memo,
                    t.balance_after,
                    t.created_at
                FROM account_transactions t
                JOIN accounts a ON t.account_id = a.id
                LEFT JOIN accounts ta ON t.target_account_id = ta.id
                WHERE t.user_id = ?
                ORDER BY t.id DESC
                """;

        return jdbcTemplate.query(sql, transactionRowMapper, userId);
    }

    public List<AccountTransaction> findAll() {
        String sql = """
                SELECT
                    t.id,
                    t.user_id,
                    t.account_id,
                    t.target_account_id,
                    a.account_number AS account_number,
                    ta.account_number AS target_account_number,
                    t.transaction_type,
                    t.amount,
                    t.memo,
                    t.balance_after,
                    t.created_at
                FROM account_transactions t
                JOIN accounts a ON t.account_id = a.id
                LEFT JOIN accounts ta ON t.target_account_id = ta.id
                ORDER BY t.id DESC
                """;

        return jdbcTemplate.query(sql, transactionRowMapper);
    }

    public void save(
            Long userId,
            Long accountId,
            Long targetAccountId,
            String transactionType,
            Long amount,
            String memo,
            Long balanceAfter
    ) {
        String sql = """
                INSERT INTO account_transactions
                (user_id, account_id, target_account_id, transaction_type, amount, memo, balance_after)
                VALUES (?, ?, ?, ?, ?, ?, ?)
                """;

        jdbcTemplate.update(
                sql,
                userId,
                accountId,
                targetAccountId,
                transactionType,
                amount,
                memo,
                balanceAfter
        );
    }
}
```

```python
%%writefile spring-lab/finance-security-app/src/main/java/com/example/demo/service/UserService.java

package com.example.demo.service;

import com.example.demo.domain.AppUser;
import com.example.demo.dto.SignupForm;
import com.example.demo.repository.UserRepository;
import org.springframework.security.crypto.password.PasswordEncoder;
import org.springframework.stereotype.Service;

@Service
public class UserService {

    private final UserRepository repository;
    private final PasswordEncoder passwordEncoder;

    public UserService(
            UserRepository repository,
            PasswordEncoder passwordEncoder
    ) {
        this.repository = repository;
        this.passwordEncoder = passwordEncoder;
    }

    public void signup(SignupForm form) {
        validateSignupForm(form);

        if (repository.findByEmail(form.getEmail()).isPresent()) {
            throw new IllegalArgumentException("이미 가입된 이메일입니다.");
        }

        String encodedPassword = passwordEncoder.encode(form.getPassword());

        repository.save(form, encodedPassword);
    }

    public AppUser findByEmail(String email) {
        return repository.findByEmail(email)
                .orElseThrow(() -> new IllegalArgumentException("사용자를 찾을 수 없습니다."));
    }

    private void validateSignupForm(SignupForm form) {
        if (form.getEmail() == null || form.getEmail().isBlank()) {
            throw new IllegalArgumentException("이메일이 필요합니다.");
        }

        if (form.getPassword() == null || form.getPassword().isBlank()) {
            throw new IllegalArgumentException("비밀번호가 필요합니다.");
        }

        if (form.getName() == null || form.getName().isBlank()) {
            throw new IllegalArgumentException("이름이 필요합니다.");
        }
    }
}
```

```python
%%writefile spring-lab/finance-security-app/src/main/java/com/example/demo/service/AccountService.java

package com.example.demo.service;

import com.example.demo.domain.Account;
import com.example.demo.dto.AccountCreateForm;
import com.example.demo.dto.DepositForm;
import com.example.demo.dto.TransferForm;
import com.example.demo.dto.WithdrawForm;
import com.example.demo.repository.AccountRepository;
import com.example.demo.repository.AccountTransactionRepository;
import org.springframework.stereotype.Service;
import org.springframework.transaction.annotation.Transactional;

import java.time.LocalDateTime;
import java.time.format.DateTimeFormatter;
import java.util.List;

@Service
public class AccountService {

    private final AccountRepository accountRepository;
    private final AccountTransactionRepository transactionRepository;

    public AccountService(
            AccountRepository accountRepository,
            AccountTransactionRepository transactionRepository
    ) {
        this.accountRepository = accountRepository;
        this.transactionRepository = transactionRepository;
    }

    public List<Account> findMyAccounts(Long userId) {
        return accountRepository.findByUserId(userId);
    }

    public void createAccount(Long userId, AccountCreateForm form) {
        if (form.getAccountName() == null || form.getAccountName().isBlank()) {
            throw new IllegalArgumentException("계좌 이름이 필요합니다.");
        }

        String accountNumber = generateAccountNumber(userId);

        accountRepository.save(userId, accountNumber, form.getAccountName());
    }

    @Transactional
    public void deposit(Long userId, DepositForm form) {
        validateAmount(form.getAmount());

        Account account = findMyActiveAccount(form.getAccountId(), userId);

        Long balanceAfter = account.getBalance() + form.getAmount();

        accountRepository.increaseBalance(account.getId(), form.getAmount());

        transactionRepository.save(
                userId,
                account.getId(),
                null,
                "DEPOSIT",
                form.getAmount(),
                form.getMemo(),
                balanceAfter
        );
    }

    @Transactional
    public void withdraw(Long userId, WithdrawForm form) {
        validateAmount(form.getAmount());

        Account account = findMyActiveAccount(form.getAccountId(), userId);

        if (account.getBalance() < form.getAmount()) {
            throw new IllegalArgumentException("잔액이 부족합니다.");
        }

        Long balanceAfter = account.getBalance() - form.getAmount();

        accountRepository.decreaseBalance(account.getId(), form.getAmount());

        transactionRepository.save(
                userId,
                account.getId(),
                null,
                "WITHDRAW",
                form.getAmount(),
                form.getMemo(),
                balanceAfter
        );
    }

    @Transactional
    public void transfer(Long userId, TransferForm form) {
        validateAmount(form.getAmount());

        if (form.getFromAccountId() == null || form.getToAccountId() == null) {
            throw new IllegalArgumentException("출금 계좌와 입금 계좌가 필요합니다.");
        }

        if (form.getFromAccountId().equals(form.getToAccountId())) {
            throw new IllegalArgumentException("같은 계좌로 이체할 수 없습니다.");
        }

        Account fromAccount = findMyActiveAccount(form.getFromAccountId(), userId);
        Account toAccount = findMyActiveAccount(form.getToAccountId(), userId);

        if (fromAccount.getBalance() < form.getAmount()) {
            throw new IllegalArgumentException("잔액이 부족합니다.");
        }

        Long fromBalanceAfter = fromAccount.getBalance() - form.getAmount();
        Long toBalanceAfter = toAccount.getBalance() + form.getAmount();

        accountRepository.decreaseBalance(fromAccount.getId(), form.getAmount());
        accountRepository.increaseBalance(toAccount.getId(), form.getAmount());

        transactionRepository.save(
                userId,
                fromAccount.getId(),
                toAccount.getId(),
                "TRANSFER_OUT",
                form.getAmount(),
                form.getMemo(),
                fromBalanceAfter
        );

        transactionRepository.save(
                userId,
                toAccount.getId(),
                fromAccount.getId(),
                "TRANSFER_IN",
                form.getAmount(),
                form.getMemo(),
                toBalanceAfter
        );
    }

    private Account findMyActiveAccount(Long accountId, Long userId) {
        if (accountId == null) {
            throw new IllegalArgumentException("계좌를 선택해 주세요.");
        }

        Account account = accountRepository.findByIdAndUserId(accountId, userId)
                .orElseThrow(() -> new IllegalArgumentException("본인 계좌만 사용할 수 있습니다."));

        if (!"ACTIVE".equals(account.getStatus())) {
            throw new IllegalArgumentException("사용할 수 없는 계좌입니다.");
        }

        return account;
    }

    private void validateAmount(Long amount) {
        if (amount == null || amount <= 0) {
            throw new IllegalArgumentException("금액은 1원 이상이어야 합니다.");
        }
    }

    private String generateAccountNumber(Long userId) {
        String timestamp = LocalDateTime.now().format(DateTimeFormatter.ofPattern("yyyyMMddHHmmss"));
        return "100-" + userId + "-" + timestamp;
    }
}
```
